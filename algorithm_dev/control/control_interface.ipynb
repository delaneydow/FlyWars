```python
# control/control_interface.py

def build_targets(tracks, track_states):
    targets = []
    for t in tracks:
        targets.append({
            "id": t.id,
            "pos": t.kf.statePost[:2].flatten(),
            "vel": t.kf.statePost[2:].flatten(),
            "state": track_states.get(t.id, "unknown")
        })
    return targets
```
